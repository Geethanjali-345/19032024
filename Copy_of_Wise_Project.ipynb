```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
%matplotlib inline
```

```python
dataset = pd.read_csv('/content/drive/MyDrive/Content/survey_lung_cancer.csv')
print("Data imported")
```

```python
dataset.head()
```

```python
dataset.tail()
```

```python
dataset.info()
```

```python
dataset.describe()
```

```python
dataset.describe(include = 'all')
```

```python
dataset = pd.get_dummies(dataset)
dataset.head()
```

```python
dataset.info() #dummyvariabletrap
```

```python
dataset = dataset.drop(columns= ['GENDER_M', 'LUNG_CANCER_NO'])
dataset.head()
```

```python
dataset = dataset.drop(columns = ['YELLOW_FINGERS', 'PEER_PRESSURE', 'CHRONIC DISEASE', 'ALCOHOL CONSUMING'])
dataset.head()
```

```python
dataset = dataset.drop(columns = ['ALLERGY '])
dataset.head()
```

```python
target = dataset['LUNG_CANCER_YES']
predictor = dataset.drop(columns = ['LUNG_CANCER_YES'])
```

```python
print(predictor.head())
print()
print(target.head())
```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
predictor = ss.fit_transform(predictor)
```

```python
predictor[5:10]
```

```python
target = target.values
target[5:10]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(predictor, target, test_size = 0.2, random_state = 0)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
lr_predicted = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(lr_predicted, y_test) * 100)
```
